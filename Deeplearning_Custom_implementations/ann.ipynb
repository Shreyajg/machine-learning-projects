### CUSTOM NEURAL NETWORK IMPLEMENTATION

```python
class Layer:
    def __init__(self,n_neurons,dim,activation):
        self.activation=activation
        self.n_neurons=n_neurons
        self.dim=dim
        self.weights=np.random.randn(n_neurons,dim)
        self.bias=np.ones(n_neurons)
    def forward(self,inp):
        self.inp=inp
        self.z=np.dot(self.weights,inp)+self.bias
        if self.activation=='ReLU':
            self.res=np.maximum(0,self.z)
        elif self.activation=='Sigmoid':
            self.res=(1/(1+np.exp(-self.z)))
        else:
            self.res=self.z
        return self.res
```

```python
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

Y = np.array([
    0,
    1,
    1,
    0
])
```

```python
class NeuralNetwork:
    def __init__(self):
        self.layers=[]
        self.layers.append(Layer(8,2,"ReLU"))
        self.layers.append(Layer(4,8,"ReLU"))
        self.layers.append(Layer(1,4,"Sigmoid"))
        self.output=0
    def forward_prop(self,inp):
        a=inp
        for layer in self.layers:
            a=layer.forward(a)
        self.a=a
        return a

    def backprop(self,lr=0.001):
        error=2*(self.a-self.output)

        for layer in reversed(self.layers):
            if layer.activation == "Sigmoid":
                error *= layer.res * (1 - layer.res)
            elif layer.activation=='ReLU':
                error=error*(layer.z>0)
            dw=np.outer(error,layer.inp)
            db=error
            prev_error=layer.weights.T @ error
            layer.weights-=lr*dw
            layer.bias-=lr*db

            error=prev_error
   
```

```python
nn=NeuralNetwork()
for i in range(1000):
    for x, y in zip(X, Y):
        pred = nn.forward_prop(x)
        e_loss=(pred-y)**2
        nn.output = y
        nn.backprop()
        print(pred)
    print(e_loss)
    print("\n")
```
